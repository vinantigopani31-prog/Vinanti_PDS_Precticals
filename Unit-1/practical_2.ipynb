```python
print("------ List ------")
l=[5,4,3,2]
print(l)
print(type(l))
print("length",len(l))
l.append(1)
print("after append",l)
l.insert(1,6)
print("after insert:",l)
l.remove(6)
print("after remove:",l)
print("count of 3:",l.count(3))
l.sort()
print("after sorting:",l)
l.clear()
print("cleared list:",l)
l=[1,2,3,4,5]
l.pop(3)
print("after pop up:",l)
l.reverse()
print("after reverse:",l)
print("length:",len(l))
print("slice",l[1:3])
print("odd no. present in the list:",l[::2])
print("last two second numbers:",l[-3:-1])
s={100,200,300}
l.extend(s)
print("after extending set:",l)
```

```python
print("------ Tuple ------")
t=(10,15,20,25,30)
print(type(t))
print("count of 50:",t.count(50))
print("index of 30:",t.index(30))
print("length",len(t))
print("slice",t[1:3])
print("odd no. present in the list:",t[::2])
print("last two second numbers",t[-3:-1])
```

```python
print("------ Set ------")
s={"a,b,c,d"}
print(type(s))
s.add("e")
print("after the addition of set:",s)
s.remove("e")
print("after the remove one ele. of set:",s)
```

```python
print("------ Dictionary ------")
d={'a':1,'b':2,'c':3,'d':4,'e':5}
print(type(d))
print(d)
print(d.keys())
print(d.values())
print(d.items())
d1=d.copy()
print("copy dictionary of d:",d1)
dn={'x':6,'y':7,'z':8}
d1.update(dn)
print("updated dictionary",d1)
print(d.get('b'))
d1.setdefault('w')
print("after set",d1)
d1.pop('w')
print("after pop",d1)
```

```python
print("------ String ------")
s1="Hello"
s2="python program"
print("Length of s1:",len(s1))
print("Length of s2:",len(s2))
print(s1.upper())
print(s2.lower())
print(s1.swapcase())
print(s2.title())
print(s2.capitalize())
print("Concatenation of s1 and s2:",s1 + "," + " " + s2 + "!")
print("count of 'm':",s2.count('m'))
print("repetition:",s1*3)
print("find:",s2.find("python"))
s3="Hello Java"
print("before strip:",s3)
print("after strip:",s3.strip())
print("Starts with python in s2:",s2.startswith("python"))
```
