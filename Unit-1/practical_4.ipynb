```python
def welcome():#no argument,no return value function
    print("Welcome to Python Programming!")

welcome()
```

```python
def display(name, age):#with argument,no return value function
    print("Name:", name)
    print("Age:", age)

display("Sahaja", 20)
```

```python
def get_n():#no argument,with return value function
  a=6
  return a
n=get_n()
print("n =", n)
```

```python
def sub(a, b):#with argument,with return value function
    return a - b

result = sub(15, 10)
print("Sub =", result)
```

```python
def interest(principal, rate=5, time=2):#fixed or default argument , with return value function
    si = (principal * rate * time) / 100
    return si

print("Interest =", interest(10000))
print("Interest =", interest(10000, 8, 3))
```

```python
def student(name, age, course):#with keyword argument value function
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)

student(course="Python", name="John", age=21)
```

```python
def total(*numbers):
    print("Numbers:", numbers)
    print("Sum =", sum(numbers))

total(10, 20, 30)
total(5, 10, 15, 20, 25)
```

```python
import math#importing math library
num = 25
print("Square Root:", math.sqrt(num))
print("Power:", math.pow(5, 3))
print("Ceiling:", math.ceil(4.2))
print("Floor:", math.floor(4.8))
print("Factorial:", math.factorial(5))
print("Value of Pi:", math.pi)
```

```python
import random#importing random library
print("Random Integer:", random.randint(1, 100))
colors = ["Red", "Blue", "Green", "Yellow"]
print("Random Choice:", random.choice(colors))
random.shuffle(colors)
print("Shuffled List:", colors)
print("Random Float:", random.random())
```

```python
import statistics#importing statistics library
data = [10, 20, 30, 40, 50]
print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode(data))
print("Standard Deviation:", statistics.stdev(data))
```
