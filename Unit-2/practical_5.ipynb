```python
# Write text file
with open("sample.txt", "w") as f:
    f.write("Hello Python\n")
    f.write("This is a text file.\n")
```

```python
# Read text file
with open("sample.txt", "r") as f:
    print("Text File:")
    print(f.read())
```

```python
# Append text file
with open("sample.txt", "a") as f:
    f.write("This line is appended.\n")

# Read again
with open("sample.txt", "r") as f:
    print("\nAfter Append:")
    print(f.read())
```

```python
import csv

# Write csv
with open("students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Roll No", "Name", "Marks"])
    writer.writerow([1, "Krupa", 95])
    writer.writerow([2, "Dhruvi", 90])
```

```python
# Read csv
with open("students.csv", "r") as f:
    reader = csv.reader(f)
    print("CSV File:")
    for row in reader:
        print(row)
```

```python
# Append csv
with open("students.csv", "a", newline="") as f:
    writer = csv.writer(f)
    writer.writerow([3, "Piya", 88])

# Read again
with open("students.csv", "r") as f:
    reader = csv.reader(f)
    print("\nAfter Append:")
    for row in reader:
        print(row)
```

```python
# Write binary data
with open("data.bin", "wb") as f:
    f.write(b"Hello Python")
```

```python
# Read binary data
with open("data.bin", "rb") as f:
    print("Binary File:")
    print(f.read())
```

```python
# Append binary data
with open("data.bin", "ab") as f:
    f.write(b" - Appended Data")

# Read again
with open("data.bin", "rb") as f:
    print("\nAfter Append:")
    print(f.read())
```
