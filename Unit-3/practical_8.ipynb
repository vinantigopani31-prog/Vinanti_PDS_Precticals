```python
import pandas as pd
```

```python
df = pd.read_csv("employee.csv")
```

```python
df.head()
```

```python
print(df.columns.tolist())
```

```python
print(df.shape)
```

```python
df.info()
```

```python
data = df['Age']
```

```python
print(data.head())
```

```python
mean = data.mean()
print("Mean:", mean)
```

```python
mode = data.mode()
print("Mode:", mode.tolist())
```

```python
variance = data.var()
print("Variance:", variance)
```

```python
minimum = data.min()
maximum = data.max()
range_value = maximum - minimum

print("Minimum:", minimum)
print("Maximum:", maximum)
print("Range:", range_value)
```

```python
Q1 = data.quantile(0.25)
Q2 = data.quantile(0.50)
Q3 = data.quantile(0.75)

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
```

```python
IQR = Q3 - Q1

print("IQR:", IQR)
```

```python
print("Mean:", data.mean())
print("Median:", data.median())
print("Mode:", data.mode().tolist())
print("Variance:", data.var())
print("Standard Deviation:", data.std())
print("Minimum:", data.min())
print("Maximum:", data.max())
print("Range:", data.max() - data.min())
print("Q1:", data.quantile(0.25))
print("Q2:", data.quantile(0.50))
print("Q3:", data.quantile(0.75))
print("IQR:", data.quantile(0.75) - data.quantile(0.25))
```
