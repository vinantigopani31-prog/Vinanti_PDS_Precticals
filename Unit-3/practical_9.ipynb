```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("employee.csv")
```

```python
print("Mean:", df['Age'].mean())
print("Median:", df['Age'].median())
print("Mode:", df['Age'].mode()[0])
print("Variance:", df['Age'].var())
print("Standard Deviation:", df['Age'].std())
print("Minimum:", df['Age'].min())
print("Maximum:", df['Age'].max())
print("Range:", df['Age'].max() - df['Age'].min())
```

```python
Q1 = df['Age'].quantile(0.25)
Q2 = df['Age'].quantile(0.50)
Q3 = df['Age'].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q2 (Median):", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Age'], bins=10)

plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.title("Distribution of Employee Age")

plt.show()
```

```python
plt.figure(figsize=(8,4))

sns.boxplot(x=df['Age'])

plt.title("Box Plot of Employee Age")
plt.xlabel("Age")

plt.show()
```

```python
#bivariate analysis
plt.figure(figsize=(8,5))

plt.scatter(df['Age'], df['MonthlyIncome'])

plt.xlabel('Age')
plt.ylabel('Monthly Income')
plt.title('Age vs Monthly Income')

plt.show()
```

```python
pd.crosstab(df['JobSatisfaction'], df['Attrition'])
```

```python
pd.crosstab(df['JobSatisfaction'], df['Attrition']).plot(kind='bar', figsize=(8,5))

plt.xlabel('Job Satisfaction')
plt.ylabel('Number of Employees')
plt.title('Job Satisfaction vs Attrition')

plt.xticks(rotation=0)
plt.show()
```

```python
pd.crosstab(df['Gender'], df['Attrition'])
```

```python
pd.crosstab(df['Gender'], df['Attrition']).plot(
    kind='bar',
    figsize=(8,5)
)

plt.xlabel('Gender')
plt.ylabel('Number of Employees')
plt.title('Gender vs Attrition')

plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='JobLevel', y='MonthlyIncome', data=df)

plt.xlabel('Job Level')
plt.ylabel('Monthly Income')
plt.title('Job Level vs Monthly Income')

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

numeric_df.head()
```

```python
correlation_matrix = numeric_df.corr()

correlation_matrix
```

```python
plt.figure(figsize=(14,10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Heatmap of Employee Dataset')
plt.show()
```
