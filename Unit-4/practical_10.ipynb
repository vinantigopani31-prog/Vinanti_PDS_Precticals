```python
#matrix multiplication
a = np.array([
    [1, 2],
    [3, 4]
])
b = np.array([
    [5, 6],
    [7, 8]
])
result = a @ b #@=np.matmul()
print(result)
```

```python
import numpy as np

# 1. Array creation
arr = np.arange(1, 13)

print("Original Array:")
print(arr)

# 2. Reshaping
arr = arr.reshape(3, 4)

print("\nReshaped Array:")
print(arr)

# 3. Indexing
print("\nElement at row 2, column 3:")
print(arr[1, 2])

# 4. Slicing
print("\nFirst two rows:")
print(arr[0:2])

print("\nFirst two columns:")
print(arr[:, 0:2])

# 5. Broadcasting
print("\nAfter adding 10:")
print(arr + 10)

# 6. Mathematical operations
print("\nArray multiplied by 2:")
print(arr * 2)

print("\nArray squared:")
print(arr ** 2)

# 7. Statistical operations
print("\nSum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
```
