```python
import pandas as pd

# Create a Series
marks = pd.Series([85, 72, 90, 68, 78])
print("Series:")
print(marks)
```

```python
# Create a DataFrame
data = {
    "Name": ["Krupa", "Vinanti", "Khushali", "Dhruvi", "Piya"],
    "Department": ["CE", "CE", "IT", "IT", "CE"],
    "Marks": [85, 72, 90, 68, 78],
    "Age": [20, 19, 21, 18, 19]
}

df = pd.DataFrame(data)

print("\nDataFrame:")
print(df)
```

```python
# Data Selection
print("\nSelected Name and Marks:")
print(df[["Name", "Marks"]])
```

```python
# Filtering
print("\nStudents with Marks >= 75:")
print(df[df["Marks"] >= 75])
```

```python
# Grouping
print("\nGrouped by Department:")
print(df.groupby("Department")["Marks"].mean())
```

```python
# Aggregation
print("\nAggregation:")
print(df.groupby("Department").agg({
    "Marks": ["mean", "max", "min"],
    "Age": "mean"
}))
```
