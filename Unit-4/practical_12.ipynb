```python
import numpy as np
from scipy import stats

data = np.array([10, 12, 15, 18, 20, 20, 22, 25, 30, 35])
```

```python
print("Mean:", stats.tmean(data))
print("Median:", np.median(data))
print("Mode:", stats.mode(data, keepdims=True).mode[0])
```

```python
print("Variance:", stats.tvar(data))
print("Standard Deviation:", stats.tstd(data))
```

```python
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
```

```python
print("\nSkewness:", stats.skew(data))
print("Kurtosis:", stats.kurtosis(data))
```

```python
# Shapiro-Wilk normality test
stat, p = stats.shapiro(data)
print("\nShapiro-Wilk Test")
print("Statistic:", stat)
print("p-value:", p)
```

```python
# One-sample t-test
stat, p = stats.ttest_1samp(data, 20)
print("\nOne-Sample t-test")
print("Statistic:", stat)
print("p-value:", p)
```
