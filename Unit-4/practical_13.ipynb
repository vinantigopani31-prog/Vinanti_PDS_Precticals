```python
import requests
from bs4 import BeautifulSoup

search_query = "headphones"
page_no = 1

while True:
    url = f"https://www.amazon.in/s?k={search_query}&page={page_no}"
    print(url)

    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                      "AppleWebKit/537.36 (KHTML, like Gecko) "
                      "Chrome/120.0.0.0 Safari/537.36",
        "Accept-Language": "en-US,en;q=0.9"
    }

    response = requests.get(url, headers=headers)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, "html.parser")

        products = soup.find_all(
            "div",
            {"data-component-type": "s-search-result"}
        )

        if not products:
            print("No products found on the page.")
            break

        for item in products:

            name_elem = item.find("h2")
            price_elem = item.find(
                "span",
                class_="a-price-whole"
            )
            discount_elem = item.find(
                "span",
                class_="savingsPercentage"
            )

            if name_elem:
                print("Product:", name_elem.get_text(strip=True))

                if price_elem:
                    print("Price:", price_elem.get_text(strip=True))
                else:
                    print("Price: Not available")

                if discount_elem:
                    print("Discount:", discount_elem.get_text(strip=True))
                else:
                    print("Discount: Not available")

                print("-" * 40)

        print("Page no:", page_no)

    else:
        print(f"Failed to fetch page. Status code: {response.status_code}")
        break

    page_no += 1
```
