```python
import numpy as np
from scipy.stats import ttest_1samp

# Sample data
marks = [72, 68, 75, 71, 69, 74, 73, 70, 76, 72]

# Expected population mean
population_mean = 70

# One-sample t-test
t_stat, p_value = ttest_1samp(marks, population_mean)

print("Sample Mean:", np.mean(marks))
print("t-statistic:", t_stat)
print("p-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
    print("There is significant difference from the population mean.")
else:
    print("Fail to reject H0")
    print("There is no significant difference from the population mean.")
```
