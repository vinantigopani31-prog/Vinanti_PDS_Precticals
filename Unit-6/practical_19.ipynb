```python
import pandas as pd

# Create DataFrame
data = {
    'Name': ['Krupa', 'Dhruvi', 'Khushali', 'Vinanti', 'Amita'],
    'Age': [20, 21, 20, 22, 21],
    'Marks': [95, 90, 78, 92, 88]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

```python
# 1. SLICING
print("\n1. Slicing:")
print(df.iloc[0:3, 0:3])
```

```python
# 2. FILTERING
print("\n2. Filtering (Marks > 85):")
filtered = df[df['Marks'] > 85]
print(filtered)
```

```python
# 3. SORTING
print("\n3. Sorting by Marks:")
sorted_df = df.sort_values(by='Marks', ascending=False)
print(sorted_df)
```

```python
# 4. CONCATENATING
data2 = {
    'Name': ['piya', 'Karni'],
    'Age': [20, 23],
    'Marks': [82, 75]
}
df2 = pd.DataFrame(data2)
print("\n4. Concatenating:")
concat_df = pd.concat([df, df2], ignore_index=True)
print(concat_df)
```

```python
# 5. AGGREGATING
print("\n5. Aggregating:")
print("Average Marks:", df['Marks'].mean())
print("Maximum Marks:", df['Marks'].max())
print("Minimum Marks:", df['Marks'].min())
print("Total Marks:", df['Marks'].sum())
```

```python
# 6. NORMALIZING
print("\n6. Normalizing Marks:")
df['Normalized_Marks'] = (
    (df['Marks'] - df['Marks'].min()) /
    (df['Marks'].max() - df['Marks'].min())
)

print(df)
```
