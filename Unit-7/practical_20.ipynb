```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, marker='s',color='green')
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["CN", "PDS", "PM", "MPI"]
marks = [75, 85, 70, 90]

plt.bar(subjects, marks,color='purple')
plt.title("Bar Chart")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90]

plt.hist(marks, bins=5)
plt.title("Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["CN", "PDS", "PM", "MPI"]
marks = [25, 30, 20, 25]

plt.pie(marks, labels=subjects, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.figure(figsize=(10, 6))

plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title("Line")

plt.subplot(2, 2, 2)
plt.bar(x, y)
plt.title("Bar")

plt.subplot(2, 2, 3)
# orientation='horizontal' turns it sideways
# edgecolor='white' adds a clean line between the connected bars
plt.hist(y, bins=8, orientation='horizontal')
plt.title("Histogram (Connected)")

plt.subplot(2, 2, 4)
plt.scatter(x, y)
plt.title("Scatter")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.figure(figsize=(10, 6))

plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title("Line")

plt.subplot(2, 2, 2)
plt.bar(x, y)
plt.title("Bar")

plt.subplot(2, 2, 3)
plt.hist(y, bins=8)
plt.title("Histogram")

plt.subplot(2, 2, 4)
plt.scatter(x, y)
plt.title("Scatter")

plt.tight_layout()
plt.show()
```
